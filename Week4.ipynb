```python
import math

def gaussian(numbers, mean, std):
    final_result = []

    for x in numbers:

      coefficient = 1 / (std * math.sqrt(2 * math.pi))
      exponent = -((x - mean) ** 2) / (2 * (std ** 2))

      result = coefficient * math.exp(exponent)

      final_result.append(result)
    return final_result

numbers = [-3, -0.4, 2]
mean = 0
std = 2

result = gaussian(numbers, mean, std)

print(result)
```

```python
import numpy as np

def gaussian_np(array, mean, std):
    coefficient = 1 / (std * np.sqrt(2 * np.pi))

    # Tính toán exponent sử dụng broadcasting của NumPy
    exponent = -((array - mean) ** 2) / (2 * (std ** 2))

    # Tính kết quả sử dụng hàm exp của NumPy
    result = coefficient * np.exp(exponent)

    return np.round(result, 2)

array = np.array([-0.1, 0.01, -3])
mean = 0
std = 3

result = gaussian_np(array, mean, std)

print(result)
```

```python
import random

def generate_dice_rolls(n, seed = 0):
    random.seed(seed)
    return [random.randint(1, 6) for _ in range(n)]

n_rolls = 1000
dice_rolls = generate_dice_rolls(n_rolls)

print(dice_rolls)
```

```python
def count_occurrences(dice_rolls, numbers):
    count = 0
    for roll in dice_rolls:
        if roll == numbers:
            count += 1
    return count

number_of_interest = 6
occurrences = count_occurrences(dice_rolls, number_of_interest)

print(occurrences)
```

```python
def calculate_probability(dice_rolls, number):
    count = 0
    for roll in dice_rolls:
        if roll == number:
            count += 1
    probability = count / len(dice_rolls)
    return round(probability, 2)

n_rolls = 1000
dice_rolls = generate_dice_rolls(n_rolls)

number_of_interest = 4

probability = calculate_probability(dice_rolls, number_of_interest)

print(probability)
```
