```python
import cv2
import numpy as np

# Load các ảnh input
bg_image1 = cv2.imread('image_w5/background1.png', 0)
ob_image = cv2.imread('image_w5/object.png', 0)

# Resize kích thước ảnh đầu vào về cùng kích thước
bg_image1 = cv2.resize(bg_image1, (640, 480))
ob_image = cv2.resize(ob_image, (640, 480))

# Chuyển đổi cân màu từ BRG sang RGB
bg_image1 = cv2.cvtColor(bg_image1, cv2.COLOR_BGR2RGB)
ob_image = cv2.cvtColor(ob_image, cv2.COLOR_BGR2RGB)

difference_three_channel = np.abs(bg_image1 - ob_image)
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('image_w5/nature.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

new_img = img + 50
new_img = np.clip(new_img, 0, 255)
new_img = new_img.astype(np.uint8)

plt.figure(figsize= (10, 10))

plt.subplot(121)
plt.imshow(img)
plt.title("Origin")

plt.subplot(122)
plt.imshow(new_img)
plt.title("Brightness")

plt.show()
```

```python
import cv2
import matplotlib . pyplot as plt
import numpy as np

# Đọc hình ảnh
img = cv2.imread ("image_w5/nature.jpg")
img = cv2.cvtColor (img, cv2.COLOR_BGR2RGB )

# Crop hình ảnh
crop_img = img[50:150, 500:700]

# Hiển thị hình ảnh
plt.figure ( figsize = (10 , 10) )

plt.subplot (121)
plt.imshow (img)
plt.title (" Original Image ")

plt.subplot (122)
plt.imshow (crop_img)
plt.title ("Cropped Image ")
```
